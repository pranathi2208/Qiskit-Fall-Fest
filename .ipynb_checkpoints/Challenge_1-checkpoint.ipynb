# Qiskit Fall Fest 2026 — Challenge 1: Predict the Output

**Goal:** Predict and compare measurement results for three one-qubit circuits. Each circuit is run for **100 shots**.

## Predictions (before running)
- **Circuit A (`X`):** expect `1` on nearly 100% of shots.
- **Circuit B (`X` then `X`):** expect `0` on nearly 100% of shots because two X gates undo each other.
- **Circuit C (`H`):** expect `0` and `1` about equally often (roughly 50% each), with small random differences.

Run the prediction first, then execute the cells and compare with the actual counts.

```python
# Run this cell if Qiskit is not installed yet
%pip install qiskit qiskit-aer matplotlib
```

```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt

simulator = AerSimulator()
SHOTS = 100
```

## Circuit A — one X gate

```python
qc_a = QuantumCircuit(1, 1)
qc_a.x(0)
qc_a.measure(0, 0)
display(qc_a.draw("mpl"))

result_a = simulator.run(transpile(qc_a, simulator), shots=SHOTS).result()
counts_a = result_a.get_counts()
print("Circuit A counts:", counts_a)
plot_histogram(counts_a)
plt.show()
```

## Circuit B — two X gates

```python
qc_b = QuantumCircuit(1, 1)
qc_b.x(0)
qc_b.x(0)
qc_b.measure(0, 0)
display(qc_b.draw("mpl"))

result_b = simulator.run(transpile(qc_b, simulator), shots=SHOTS).result()
counts_b = result_b.get_counts()
print("Circuit B counts:", counts_b)
plot_histogram(counts_b)
plt.show()
```

## Circuit C — one H gate

```python
qc_c = QuantumCircuit(1, 1)
qc_c.h(0)
qc_c.measure(0, 0)
display(qc_c.draw("mpl"))

result_c = simulator.run(transpile(qc_c, simulator), shots=SHOTS).result()
counts_c = result_c.get_counts()
print("Circuit C counts:", counts_c)
plot_histogram(counts_c)
plt.show()
```

## Compare the results

Use the counts and histograms above to complete the comparison:
- Circuit A should produce `1` every time because X flips `|0⟩` to `|1⟩`.
- Circuit B should produce `0` every time because the second X flips the qubit back.
- Circuit C should produce both `0` and `1` with approximately equal frequency because H creates an equal superposition. Individual 100-shot results can vary.

**Challenge question:** A quantum gate changes a qubit's state. When measured, the probabilities of `0` and `1` depend on that state. X flips the basis state, two X gates cancel, and H creates equal probabilities for the two measurement outcomes.
